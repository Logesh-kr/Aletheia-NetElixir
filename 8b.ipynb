```python
from itertools import permutations

INF = float('inf')

# 6-city road distance matrix (in km)
cost = [
    [INF, 10, 15, 20, 18, 25],
    [10, INF, 35, 25, 17, 30],
    [15, 35, INF, 30, 28, 20],
    [20, 25, 30, INF, 16, 18],
    [18, 17, 28, 16, INF, 14],
    [25, 30, 20, 18, 14, INF]
]

cities = ['A', 'B', 'C', 'D', 'E', 'F']
n = len(cost)

# ---------- Brute Force ----------
def tsp_brute_force(cost):
    cities_idx = list(range(1, n))
    best_cost = INF
    best_path = None

    for perm in permutations(cities_idx):
        path = [0] + list(perm) + [0]
        total = 0

        for i in range(n):
            total += cost[path[i]][path[i + 1]]

        if total < best_cost:
            best_cost = total
            best_path = path

    return best_path, best_cost


# ---------- Nearest Neighbour ----------
def nearest_neighbour(cost):
    visited = [False] * n
    path = [0]
    visited[0] = True
    total = 0
    current = 0

    for _ in range(n - 1):
        nearest = -1
        min_cost = INF

        for city in range(n):
            if not visited[city] and cost[current][city] < min_cost:
                min_cost = cost[current][city]
                nearest = city

        path.append(nearest)
        visited[nearest] = True
        total += min_cost
        current = nearest

    path.append(0)
    total += cost[current][0]

    return path, total


# ---------- Display Cost Matrix ----------
print("6-City Cost Matrix (km)\n")

print("     ", end="")
for c in cities:
    print(f"{c:>5}", end="")
print()

for i in range(n):
    print(f"{cities[i]:>3} ", end="")
    for j in range(n):
        if cost[i][j] == INF:
            print(f"{'INF':>5}", end="")
        else:
            print(f"{cost[i][j]:>5}", end="")
    print()

# ---------- Solve ----------
bf_path, bf_cost = tsp_brute_force(cost)
nn_path, nn_cost = nearest_neighbour(cost)

print("\nBrute Force Solution")
print("--------------------")
print("Route :", " -> ".join(cities[i] for i in bf_path))
print("Cost  :", bf_cost, "km")

print("\nNearest Neighbour Solution")
print("--------------------------")
print("Route :", " -> ".join(cities[i] for i in nn_path))
print("Cost  :", nn_cost, "km")

print("\nComparison")
print("----------")
print("Optimal Cost          :", bf_cost, "km")
print("Nearest Neighbour Cost:", nn_cost, "km")

if nn_cost == bf_cost:
    print("Nearest Neighbour found the optimal solution.")
else:
    print("Nearest Neighbour is",
          nn_cost - bf_cost,
          "km costlier than the optimal solution.")
```
